### practise I: Solar Panel Analysis
We received a raw data stream from the main probe, representing the efficiency of individual cells of the solar array. The data arrived in one long series, but the solar array is actually a grid.

**Assignment:**
1.  Create an array (`solar_stream`) containing **20 random integers** in the range 0 to 100 (representing % efficiency).
2.  Reshape this stream into a matrix (grid) with dimensions **4 rows and 5 columns**.
3.  Calculate and print the **average efficiency** of the entire array.
4.  Find and display the **maximum value** in the grid (the most efficient cell).

```python
import numpy as np

# 1. Generate random data (20 data points)
solar_stream = np.random.randint(0, 101, size=20)
print("Raw Stream:\n", solar_stream)

# 2. Reshape to 4x5 grid
solar_grid = solar_stream.reshape(4, 5)
print("\nSolar Grid (4x5):\n", solar_grid)

# 3. & 4. Statistics
print(f"\nAverage Efficiency: {np.mean(solar_grid):.2f}%")
print(f"Max Efficiency: {np.max(solar_grid)}%")
```

### practise II: Radiation Anomaly Detection (Vectorization & Filtering)
Sensors on the ship's hull measure radiation levels. Values are in "safe units". We must perform re-calibration and immediately isolate critical values.

**Assignment:**
1.  Create an array (`raw_readings`) containing numbers from **10 to 50** (inclusive) with a step of **5**.
2.  **Vectorization:** Sensor calibration after a solar storm requires multiplying all values by **3.5**. Do this without using a loop. Save results to the variable `calibrated_readings`.
3.  **Filtration:** Print only those calibrated values that are **higher than 100** (critical level).

```python
import numpy as np

# 1. Create sequence 10, 15, 20 ... 50
raw_readings = np.arange(10, 51, 5)
print("Raw Readings:", raw_readings)

# 2. Vectorized calibration
calibrated_readings = raw_readings * 3.5
print("Calibrated data:", calibrated_readings)

# 3. Filtering anomalies
anomalies = calibrated_readings[calibrated_readings > 100]
print("CRITICAL LEVELS DETECTED:", anomalies)
```

## Homework (Project: Mars)
We are commencing primary data collection from the surface of Mars.

**Mission:** Our task is to process the first batch of telemetry from the landing module using NumPy.

**Goal:** Create a script in `mars_analaysis_01.ipynb` that will simulate and analyze data about the atmosphere and terrain.

---

### Assignment:

**1. Atmospheric Log (Data Creation):**
* Create a timeline (`time_log`) for the first **24 hours** of measurement (values 0 to 23).
* We will simulate temperature measurement (`temp_readings`) - create an array of 24 random numbers in the range **-100 to -20 °C**.

**2. Topographic Scan (Reshape):**
* The Rover sent a raw data stream from LiDAR (elevation map). **100 numbers** arrived (random values 0 to 500 meters) in one row.
* Save this data to the variable `raw_terrain`.
* Reformat this stream into a **sector map (matrix)** of size **10x10**.

**3. Environment Analysis (Statistics):**
* Calculate and print the **average temperature** for the last 24 hours.
* Find and print the **highest point** in the scanned sector (max value in the terrain map).

**4. Sensor Calibration (Vectorization):**
* We discovered that the temperature sensor has a systematic error and shows **5 degrees more** than reality.
* Perform a correction of the entire `temp_readings` array (subtract 5) - without using a loop.
* Print the first 5 values of the corrected array.

```python
import numpy as np

# --- 1. Atmospheric Log ---
# Time axis (0-23 hours)
time_log = np.arange(24)
print("Time Log:", time_log)

# Temperature simulation (-100 to -20 C)
temp_readings = np.random.randint(-100, -19, size=24)
print("Raw Temps:", temp_readings)

# --- 2. Topographic Scan ---
# 100 raw data points (0-500 meters elevation)
raw_terrain = np.random.randint(0, 501, size=100)

# Reshape to 10x10 Grid
terrain_map = raw_terrain.reshape(10, 10)
print("\nTerrain Map (10x10):\n", terrain_map)

# --- 3. Environmental Analysis ---
# Average Temperature
avg_temp = np.mean(temp_readings)
print(f"\nAverage Daily Temp: {avg_temp:.2f} C")

# Highest Peak
max_elevation = np.max(terrain_map)
print(f"Highest Peak Detected: {max_elevation} m")

# --- 4. Sensor Calibration ---
# Vectorized correction (-5 degrees)
corrected_temps = temp_readings - 5
print("\nCorrected Temps (First 5):", corrected_temps[:5])
```
