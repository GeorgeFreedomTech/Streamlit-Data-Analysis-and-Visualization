## 0.1. Jupyter Notebook (Our Working Tool)

Jupyter Notebook (`.ipynb`) is the standard tool for data science and signal analysis. It allows us to combine:

1. **Live code:** we can immediately run and test our code

2. **Formatted text (Markdown):** for explaining ideas, our notes, and measurement reports

3. **Visualization:** charts and tables are displayed directly below the code

Unlike classic scripts that run as a whole, Jupyter works in **cells**. This allows us to perform experiments "independently" at different points in the analysis.

### ⚠️ 0.1.1. Protocol: Code Chronology

* in standard Python programming (`.py` file), code is executed strictly from top to bottom

* in Jupyter (`.ipynb` file), we have the freedom to run selected cells `independently and in any order` - this brings risks along with benefits

### 0.1.2. Python Interpreter x Jupyter Kernel

- `interpreter` **executes** our python code = for Python we need a Python interpreter
- `kernel` **wraps** the selected interpreter (e.g., the Python one) and enables a specific way of executing code in Jupyter Notebook (JPN)

**What to watch out for:**
* the kernel **sees** only variables and functions that we **have run**

* if we work with a variable or function that you haven't run yet (or we deleted it and restarted the kernel), the program won't recognize it, and we get an error

* **Recommendation:** Keep your notebook in a state where it is always executable linearly from top to bottom = (the Run All button runs all cells, and thus all variables and functions get into the memory / consciousness of the kernel).

## 0.3. JPN Environment Preparation (Setup)

To analyze samples and sensor data, we need an isolated environment.

### Installation Procedure (Terminal):
**1. Creating a virtual environment (venv):**
This is usually 1-2 versions older than the current Python version.

- we will have multiple global Python versions, from which we will then choose according to project needs

```bash
# Windows (example for Python 3.12)
py -3.12 -m venv data_venv

# MacOS/Linux
python3 -m venv data_venv
```
2. Environment activation:

```bash
Windows: data_venv\Scripts\activate

Mac/Linux: source data_venv/bin/activate
```
- Troubleshooting (Win):
```bash
Windows: ./activate
```

3. Library installation: We install the complete set of tools for analysis and visualization at once (in the active venv).

```bash
pip install numpy pandas matplotlib ipykernel streamlit
```

4. In the JPN file, select the python interpreter ("python.exe") from our venv as the kernel (top right button "Select kernel")

- Troubleshooting (Jupyter Kernel): 
    - If Jupyter in VS Code doesn't see your virtual environment, we must `register it manually`:

        - first, we activate our venv

        - we perform the venv registration

```bash
python -m ipykernel install --user --name=data_venv --display-name "Python - Data Science VENV"
```
p.s. for "name" we use our selected venv (here "data_venv"), for "display-name" the name under which we want to display our venv in the kernel menu (here "Python - Data Science VENV")

## 0.4. Jupyter Environment - Basic Controls
In JPN, we use two basic types of cells - **active** "Code" cells and **passive** "Markdown" cells.

1. `**Markdown:**` Serves for text, notes, explanations, comments, and documentation of our projects, lectures, and materials

2. `**Code:**` Serves for writing and running (Python) code - i.e., analysis, calculations, rendering... etc.

### 0.4.1 Working with MARKDOWN fields
We structure text for clarity, similar to a technical report:

- `Headings:` Hashtags # for Main Heading (1st level), ## Subheading (2nd level), etc.

- `Lists:` Use dashes - for bullets or 1. for a numbered list.

- `Highlighting:` use asterisks ** around **what we want to highlight**, or "backtick" for `contrast` highlighting

- `Editing:` Double-clicking a text cell switches you to editing mode.

- `Confirming:` Shift + Enter or Ctrl + Enter renders the text (or runs the code).

### 0.4.1 Working with CODE fields

We write code into cells marked **Code.**

- `Running:` The ▶️ icon on the left of the cell or the keyboard shortcut Shift + Enter.

- `Output:` The result of the operation is displayed immediately below the cell.

- `Reset:` Right mouse button -> Clear Cell Output (clears output - executed code)

## 0.5. Tool Introduction (Tech-Stack)
We will use four specialized libraries for data processing. While using them, we will familiarize ourselves with the `ETL process` (Extract, Transform, Load). We will acquire, manipulate/transform data (and use it to create visual representations), and finally, we can store it permanently (e.g., in a file or DB).

### 0.5.1 `NumPy:` Structural Foundation

Sensor data comes in various forms - often as a **raw stream** of numbers (records) or **loosely structured** data.

- **Principle:** `Low-level engine for numerical data`. It allows us to "reshape" the data. We can imagine a continuous stream of data (values from a sensor). NumPy allows us to transform this 1D stream into a `well-structured and organized` 2D table (matrix) - e.g., a stream of 100 records into a 4x25 table/grid.

- **Usage:** we can work even with raw data and prepare it for later processing at a higher level using Pandas

### 0.5.2 `Pandas:` Analytical Engine (ETL)

Once we need to load data from CSV, Excel, JSON, or a database, we use Pandas. `Data extraction` is the first part of the ETL process and is only a minor part of Pandas' capabilities. Its true power lies in its abilities to subsequently `transform, sort, and rearrange data` (the second part of the ETL process).

- **Principle:** we can imagine Pandas as a pumped-up programmable Excel on steroids

- **Scenario:** we receive a massive telemetry log (voltage, vibrations, time, temperature, sensor ID, material composition) - that is a lot of things, and most of them are not useful for our task = useless ballast / noise

- **Usage:** with the help of Pandas, we clean the data, filter the relevant signal / data (e.g., voltage fluctuation and measured temperature), and further, we can sort, analyze, group (e.g., by sensor ID) or compare results on a timeline, etc.

### 0.5.3 `Matplotlib:` Vizualisation (Recon)

Humans process most information visually. Mere numbers or data sets don't tell us much.

- **Principle:** creation and setting of detailed `static charts`

- **Usage:** ideal for creating overviews and `quick "snapshots" of the situation` (what the data says). We set axes, labels, scales, and the type of chart (line, scatter, on a map background, etc.)

### 0.5.4 `Streamlit:` Interactive Dashboard

To truly understand the data, we need the ability to interact with it.

- **Principle:** allows converting our Python scripts into a `dynamic web application` without knowledge of HTML/CSS

- **Usage:** zooming, filtering (switching data sets - e.g., daily overview or monthly summary), setting various parameters (e.g., consumption, temperature, composition..)

- **Result:** `interactive dashboard accessible online` - we can make changes to parameters, scales, comparisons, and observe the impacts of changes
