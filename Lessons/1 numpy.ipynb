**NumPy** (Numerical Python) is the `cornerstone` of scientific computing.

## 1.1 Creating Data Arrays

We rarely enter sensor data manually. We often `generate` it manually or load it `as streams`.

### 1.1.1 Manual vs. Automatic Creation

- when creating an **array**, its `data type` and `size` must be determined
- if the data type is not determined, numpy will try to estimate the data type based on the entered values in the array
- data type is determined using dtype - e.g.:
    * int8 - 8-bit integer (-128 to 127)
    * int16 - 16-bit integer (-32,768 to 32,767)
    * int32 - 32-bit integer (-2,147,483,648 to 2,147,483,647) - **standard**
- arrays can also have other data types (str, float, bool ..)

```python
# 1. creating arrays

# 1D array - basic one-dimensional array
# standard Python list vs NumPy Array
raw_list = [1, 2, 3, 4, 5]
sensor_stream = np.array(raw_list, dtype=np.int32)

print("Sensor Stream:", sensor_stream)
print("Type:", type(sensor_stream))

# 2D array - two-dimensional array
gamma_detection = np.array([[100, 315, 3], [62, 599, 21]], dtype=np.int32) # nested list = matrix / table
print("Detected gamma energy: ", gamma_detection)

# it is also possible to create multidimensional arrays 3D, 4D ..

# we can use an existing collection to create an array
array_from_list = np.array([7, 8, 9]) # from listu
array_from_tuple = np.array((10, 11, 12)) # from tuple
print("Backup sensor results:", array_from_list)
print("Backup security logs:", array_from_tuple)
```

```python
# 2. calibration - arrays with pre-filled values
# suitable for array initialization

# array with zeros
calibration_grid = np.zeros((3, 4), dtype=np.int32) # 2D array: 3 rows 4 columns
print("\nCalibration Grid (Zeros):\n", calibration_grid)

# array with ones
signal_boost = np.ones((2, 3)) # 2D array: 2 rows 3 columns
print("\nSignal Boost (Ones):\n", signal_boost)

# creating an array with a sequence
# FROM-TO with a step: using arange(start, stop, step) - similar to range in Python
time_log = np.arange(0, 60, 5) # 0 to 60 seconds, step 5s
print("\nTime Log:", time_log)

# FROM-TO with a number of points: using linspace(start, stop, number_of_points)
# ideal for graphs - creates "points" with the same "distance" between them in the given range
flight_path = np.linspace(0, 100, 5) 
print("\nFlight Bot Asteroid_Path Segments:", flight_path)
```

## 1.2. Exploration and Shape (Shape & Reshape)

Data often arrives as a long, unstructured **stream / sequence (1D stream)**. To find patterns in them, we often have to `rearrange` them into a table or matrix.

Let's say the probe sends data row by row, but we receive it as one long series of values:

```python
# raw data stream from a thermal camera (12 pixels)
raw_thermal_data = np.array([22, 24, 25, 28, 30, 32, 31, 29, 25, 23, 21, 20])

# shape - shape of the array (number of dimensions and their size)
print(f"Shape: {raw_thermal_data.shape}") # (12,) - 1D array, 12 elements
print(f"Dimensions: {raw_thermal_data.ndim}") # number of dimensions (0-point, 1-row, 2-table..)
print(f"Data Type: {raw_thermal_data.dtype}") # data type (int32, int64 ..)

# RESHAPE: transformation of data "stream" into a 3x4 grid / table (thermal grid)
# - the order of elements remains the same
# - after reshape, there must be the same number of elements as before
thermal_grid = raw_thermal_data.reshape(3, 4) # 3 rows, 4 columns (= 12 elements)

print("\nThermal Grid (3x4):\n", thermal_grid)
print(f"New Shape: {thermal_grid.shape}") # (3, 4) - 2D array with 3 rows and 4 columns = 12 elements
```

## 1.3. Mathematics and Vectorization (Key Feature!)

This is the reason why using NumPy is worth it. If we want to perform an operation for all numbers in a Python list, we have to write a for loop. In NumPy, we do it **all at once**. It is like giving one order to the entire reconnaissance unit instead of each member individually.

```python
# Analyzing the thermal grid

print(f"Max Temperature: {np.max(thermal_grid)} C") # max value
print(f"Min Temperature: {np.min(thermal_grid)} C") # min value
print(f"Average Temperature: {np.mean(thermal_grid):.2f} C") # average
print(f"Standard Deviation: {np.std(thermal_grid):.2f}") # standard deviation
```

### 1.3.1. Vector Operations (Broadcasting)
Let's imagine we need to **calibrate** sensors – add 2 degrees to all values.

```python
# Python (Slow loop = DON'T DO THIS with large data)
# for i in range(len(data)): data[i] += 2

# NumPy (Vectorization = fast and clean work)
calibrated_grid = thermal_grid + 2

print("Original:\n", thermal_grid)
print("\nCalibrated (+2):\n", calibrated_grid)

# logical and comparison methods with boolean result (True/False)
print("\nIs temperature critical (> 30)?\n", calibrated_grid > 30)
```

## 1.4. Indexing and Slicing (Targeting Data)

We need to surgically target and extract **specific data** from our matrix. NumPy uses the format [row, column] - `index position of the element`.

```python
sector_map = np.array([
    [10, 11, 12],
    [20, 21, 22],
    [30, 31, 32]
])

# 1. specific element [row, column]
print("Element at [0,0]:", sector_map[0, 0])

# 2. slicing - we get the second row
print("Second row (Sector 2):", sector_map[1, :])

# 3. slicing - we get the last column
print("Last column (Sensor C):", sector_map[:, 2])
```

### 1.4.1. View vs. Copy (Data Safety)

Watch out for memory management.

`View:` **Look** at the data. `Changing the View reflects in the original too`. Efficient and fast memory work - we don't create new data, we just adjust the **view** and **scope** according to what interests us .. the data is still the same

`Copy:` A real clone of the data. `Safe for experiments` = when we want to work with data independently, without affecting the original data

```python
# creating a VIEW - SLICING in NP (unlike Python lists) creates a VIEW automatically!
sub_sector = sector_map[0:2, 0:2]
print("View BEFORE modification:\n", sub_sector)

# modifying the view
sub_sector[0, 0] = 999

print("Modified View:\n", sub_sector)
print("\nOriginal Map (Affected!):\n", sector_map)
# 999 will appear in the original too

# creating a COPY
delta_sector = sector_map.copy()
```

## 1.5 Filtration (Searching for Anomalies)

How to quickly find values that exceed limits in a data stream?

```python
telemetry = np.array([100, 102, 98, 150, 101, 99, 450, 100])

# direct filtering - returns elements that meet the condition
anomalies = telemetry[telemetry > 120]
print("Detected Anomalies:", anomalies)

# using the "where" condition - returns indices of elements meeting the condition
anomaly_indices = np.where(telemetry > 120) # results in the form of a tuple
print("Indices of anomalies:", anomaly_indices)
print(telemetry[3])
print(telemetry[6])
```
